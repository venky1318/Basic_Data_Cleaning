```python
# Task 25: Basic Data Cleaning

## Objective
Perform basic data cleaning by handling missing values,
removing duplicate records, and correcting data types.

## Deliverables
- Original dataset
- Cleaned dataset
- Duplicate identification and removal
- Missing-value handling
- Data-type correction
- Before-and-after comparison
- Explanation of cleaning steps
```

```python
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
data = {
    "student_id": [
        "1001", "1002", "1003", "1004",
        "1005", "1006", "1007", "1008",
        "1009", "1010", "1003", "1007"
    ],

    "student_name": [
        "Aarav", "Diya", "Kabir", "Ananya",
        "Ishaan", "Meera", "Rohan", "Sara",
        "Vivaan", "Aditi", "Kabir", "Rohan"
    ],

    "department": [
        "Computer Science", "Electronics", "Mechanical",
        "Computer Science", np.nan, "Electronics",
        "Mechanical", "Computer Science", "Civil",
        "Electronics", "Mechanical", "Mechanical"
    ],

    "age": [
        "20", "21", "twenty_two", "22",
        "21", "20", "23", "21", "20", "22",
        "twenty_two", "23"
    ],

    "marks": [
        "86", "91", "74", np.nan,
        "68", "88", "79", "93", "72", "84",
        "74", "79"
    ],

    "attendance": [
        "92", "88", "79", "96",
        "72", np.nan, "85", "94", "80", "86",
        "79", "85"
    ],

    "joining_date": [
        "2023-01-15", "2022-07-10", "2021-03-22",
        "2023-09-05", "2020-11-18", "2022-02-14",
        "2021-08-30", "2024-01-08", "2022-05-16",
        "2023-04-12", "2021-03-22", "2021-08-30"
    ]
}

df = pd.DataFrame(data)

print("Original dataset created successfully!")

print("Total rows:", df.shape[0])
print("Total columns:", df.shape[1])

display(df)
```

```python
df.to_csv("student_data_raw.csv", index=False)

print("Original dataset exported successfully!")
```

```python
df = pd.read_csv(
    "student_data_raw.csv",
    dtype="string"
)

print("Original CSV loaded successfully!")

display(df)
```

```python
print("ORIGINAL DATASET DIMENSIONS")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Shape:", df.shape)
```

```python
print("ORIGINAL DATA TYPES")

display(df.dtypes.to_frame(
    name="Original Data Type"
))
```

```python
print("INITIAL MISSING VALUE COUNTS")

display(df.isnull().sum())
```

```python
print("DUPLICATE RECORD ANALYSIS")

duplicate_count = df.duplicated().sum()

print("Total duplicate records:", duplicate_count)

duplicate_rows = df[
    df.duplicated(keep=False)
]

display(duplicate_rows)
```

```python
before_duplicate_removal = len(df)

df_cleaned = df.drop_duplicates().copy()

after_duplicate_removal = len(df_cleaned)

duplicates_removed = (
    before_duplicate_removal - after_duplicate_removal
)

print("BEFORE DUPLICATE REMOVAL:", before_duplicate_removal)
print("AFTER DUPLICATE REMOVAL:", after_duplicate_removal)
print("DUPLICATES REMOVED:", duplicates_removed)

display(df_cleaned)
```

```python
duplicate_rows.to_csv(
    "duplicate_records.csv",
    index=False
)

print("Duplicate records saved successfully!")
```

```python
df_cleaned["age"] = pd.to_numeric(
    df_cleaned["age"],
    errors="coerce"
)

df_cleaned["marks"] = pd.to_numeric(
    df_cleaned["marks"],
    errors="coerce"
)

df_cleaned["attendance"] = pd.to_numeric(
    df_cleaned["attendance"],
    errors="coerce"
)

print("Numerical columns converted successfully!")

display(
    df_cleaned[
        ["student_name", "age", "marks", "attendance"]
    ]
)
```

```python
df_cleaned["joining_date"] = pd.to_datetime(
    df_cleaned["joining_date"],
    errors="coerce"
)

print("Joining date converted into datetime format.")

display(
    df_cleaned[
        ["student_name", "joining_date"]
    ]
)
```

```python
print("DATA TYPES AFTER CORRECTION")

display(
    df_cleaned.dtypes.to_frame(
        name="Corrected Data Type"
    )
)
```

```python
print("MISSING VALUE COUNTS AFTER TYPE CORRECTION")

missing_counts = df_cleaned.isnull().sum()

display(
    missing_counts.to_frame(
        name="Missing Count"
    )
)

print(
    "\nTotal missing cells:",
    df_cleaned.isnull().sum().sum()
)
```

```python
missing_percentage = (
    df_cleaned.isnull().sum() / len(df_cleaned)
) * 100

print("MISSING VALUE PERCENTAGES")

display(
    missing_percentage.round(2).to_frame(
        name="Missing Percentage"
    )
)
```

```python
numerical_columns = [
    "age",
    "marks",
    "attendance"
]

for column in numerical_columns:
    median_value = df_cleaned[column].median()

    df_cleaned[column] = (
        df_cleaned[column].fillna(median_value)
    )

    print(
        f"{column}: Missing values filled with median = "
        f"{median_value}"
    )
```

```python
df_cleaned["department"] = (
    df_cleaned["department"].fillna("Unknown")
)

print("Missing department values handled.")

display(
    df_cleaned[["student_name", "department"]]
)
```

```python
# Convert numerical columns into integer types
df_cleaned["age"] = df_cleaned["age"].astype("int64")
df_cleaned["marks"] = df_cleaned["marks"].astype("int64")
df_cleaned["attendance"] = (
    df_cleaned["attendance"].astype("int64")
)

# Convert department into a categorical type
df_cleaned["department"] = (
    df_cleaned["department"].astype("category")
)

print("Final data types corrected successfully!")

display(df_cleaned.dtypes.to_frame(
    name="Final Data Type"
))
```

```python
print("FINAL MISSING VALUE CHECK")

display(df_cleaned.isnull().sum())

print(
    "Total remaining missing values:",
    df_cleaned.isnull().sum().sum()
)
```

```python
print("FINAL DUPLICATE CHECK")

remaining_duplicates = df_cleaned.duplicated().sum()

print("Remaining duplicate records:", remaining_duplicates)
```

```python
print("FINAL CLEANED STUDENT DATASET")

print("Total cleaned records:", len(df_cleaned))

display(df_cleaned)
```

```python
cleaning_summary = pd.DataFrame({
    "Metric": [
        "Original records",
        "Duplicate records removed",
        "Records after duplicate removal",
        "Missing cells after type correction",
        "Missing cells after treatment",
        "Remaining duplicate records",
        "Final cleaned records"
    ],

    "Value": [
        len(df),
        duplicates_removed,
        after_duplicate_removal,
        4,
        df_cleaned.isnull().sum().sum(),
        df_cleaned.duplicated().sum(),
        len(df_cleaned)
    ]
})

display(cleaning_summary)
```

```python
dtype_comparison = pd.DataFrame({
    "Original Data Type": df.dtypes.astype(str),
    "Cleaned Data Type": df_cleaned.dtypes.astype(str)
})

display(dtype_comparison)
```

```python
df_cleaned.to_csv(
    "student_data_cleaned.csv",
    index=False
)

cleaning_summary.to_csv(
    "cleaning_summary.csv",
    index=False
)

print("Cleaned dataset exported successfully!")
print("Cleaning summary exported successfully!")
```
